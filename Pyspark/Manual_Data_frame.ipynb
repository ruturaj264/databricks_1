```python
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, FloatType, DateType, DecimalType
from pyspark.sql.functions import expr, lit, col

spark = SparkSession.builder.appName("MyApp2").getOrCreate()
```

```python
my_data = [
            (1,"Smith",36636.00,"M",3000),
            (2,"Rose",40288.00,"M",1000),
            (3,"Williams",42114.00,"M",7000),
            (4,"Anne",39192.00,"F",4000),
            (5,"Mary",12345.00,"F",6000)
        ]

my_columns = ['id', 'name', 'amount', 'gender', 'salary']

df = spark.createDataFrame(my_data, my_columns)
df.display()
df.printSchema()
```
